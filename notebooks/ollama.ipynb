## 1. Basic Functionality

```python
from enterprise_ai.llm.providers.ollama import OllamaProvider
from enterprise_ai.schema import Message
```

```python
# Initialize provider
provider = OllamaProvider(model_name="granite3.2-vision")  # Use a model you have downloaded

# Test model info detection
model_info = provider.get_model_info()
print(f"Model: {model_info.id}")
print(f"Context window: {model_info.context_window}")
print(f"Features: {model_info.features}")

# Test basic completion
messages = [Message.user_message("Tell me a short joke about programming")]
response = provider.complete(messages)
print(f"Response: {response.content}")
```
